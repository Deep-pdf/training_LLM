```python
import numpy as np
```

```python
#Training Data
x=np.array([1,2,3,4,5],dtype=float)
y=np.array([35,42,58,66,72], dtype=float)
```

```python
#parameters
b0= 0
b1= 0
```

```python
#learning rate
learning_rate = 0.01
```

```python
#number of training iterations
epochs = 1000
```

```python
n=len(x)
for i in range(epochs):
  #predictions
  y_pred= b0+b1 *x

  #calculate gradients
  b0_grad=(-2/n)*np.sum(y-y_pred)
  b1_grad= (-2/n)*np.sum(x*(y-y_pred))

  #update parameters
  b0= b0 -learning_rate * b0_grad
  b1= b1 -learning_rate * b1_grad
```

```python
print("Intercept: ", b0)
print("Slope: ", b1)
```

```python
#prediction
x_new= 6
pred = b0 + b1 * x_new
print("prediction: ", pred)
```
